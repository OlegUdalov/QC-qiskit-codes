# CZ decomposition

Here I decompose entangling CZ gate into a combination (sum) of 4 single qubit gates. The decomposition formula is taken from: 

arXiv:1909.07534v2 [quant-ph] 6 Jan 2021



It means that CZ = 1/4 * (I + Z_1 + Z_2 - Z_1 x Z_2).

One needs to perform measurements on four different quantum circuits and then process the measurements results using classical comupting. Using this approach one can calculate some average values for CZ usin single qubit gates only. 

This approach allows to perform so-called space-cuts of quantum circuits.
