```python
import numpy as np
import time
import sys
import math

import matplotlib.pyplot as plt

import qiskit
from qiskit import QuantumRegister as Q_R
from qiskit import ClassicalRegister as C_R
from qiskit_aer import Aer
from qiskit_aer import AerSimulator
from qiskit import QuantumCircuit, transpile

import bluequbit

import qiskit.qasm2

```

```python
#q = qiskit.qasm2.load("P1_little_peak.qasm")
#q = qiskit.qasm2.load("P3__sharp_peak.qasm")
q = qiskit.qasm2.load("P4_golden_mountain.qasm")
#q.measure_all()
#q.draw('mpl')
```

```python
i_ksi = 600
start_time = time.time()
#1 Direct circuit simulation
#simulator = AerSimulator()
#qc_merged_t = transpile(q, simulator)
#result = simulator.run(qc_merged_t,shots = 10000).result()
#counts = result.get_counts(qc_merged_t)
#top_10_direct = sorted(counts.items(), key=lambda item: item[1], reverse=True)[:10]
#end_time = time.time()
#elapsed_time_direct = end_time - start_time

#2 MPS circuit simulation
start_time = time.time()
simulator = AerSimulator(method='matrix_product_state',
    enable_truncation = True,
    zero_threshold = None,
    validation_threshold = None,
    matrix_product_state_max_bond_dimension = i_ksi,
    matrix_product_state_truncation_threshold = None,
    chop_threshold = None
    )
tcirc = transpile(q, simulator)
result = simulator.run(tcirc,shots = 1000000).result()
counts = result.get_counts(0)
top_10_MPS = sorted(counts.items(), key=lambda item: item[1], reverse=True)[:10]
end_time = time.time()
elapsed_time_MPS= end_time - start_time

# Display side-by-side
print(f"max_bond_dimension: {i_ksi}")
print(f"{'Dict1 (key=value)':<20} | {'Dict2 (key=value)':<20}")
print("-" * 43)
for (k1, v1), (k2, v2) in zip(top_10_MPS, top_10_MPS):
    print(f"{k1} = {v1:<16} | {k2} = {v2}")

#print(f"\nDirect simulation time: {elapsed_time_direct}")
print(f"MPS simulation time: {elapsed_time_MPS} \n")
```
